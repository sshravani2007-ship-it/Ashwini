Abstraction

It becomes blueprint for other classes

simple interface + hidden machinery

```python
from abc import ABC,abstractmethod #import tools for abstract class
class Animal(ABC): #Create an abstract parent class

  @abstractmethod  #Require child classes to implement sound()
  def sound(self): #Define the require method
    pass           #No actual animal sound is defined here

class Dog(Animal):  #Dog inherits rule from Animal
  def sound(self):   #Dog provides his version of sound()
    print("Woof!")  #Dog's implementation

class Cat(Animal):   #cat inherits the rule from Animal
  def sound(self):   #Cat provides its version of sound()
    print("Meow!")   #cats implementation

dog = Dog()           #Create a dog object
cat = Cat()           #create a cat object

dog.sound()          #ask dog to make its sound
cat.sound()           #ask cat to make its sound
```

```python
from abc import ABC,abstractmethod #import abstraction class
class Shape(ABC):   #create an abstract parent called shape

  @abstractmethod #Require every concrete shape to implement area()
  def area(self): #Define the common operation: area
    pass           #shape doesn't choose one formula

class Circle(Shape):  #Circle follows the shape rule
  def area(self):    #circle provides its own area implementation
    print("I calculate the area of a circle") #circle's behaviour
class Square(Shape):  #square follows the shape rule
  def area(self): #square provides its own area implementation
    print("I calculate the area of a square") #squares behaviour
circle = Circle()  #create a circle object
square = Square()  #create a square object
circle.area()   #use circles implementation
square.area()    #use squares implementation



```

```python
from abc import ABC,abstractmethod #import abstract-class tools
class Shape(ABC): #shape is an abstract base class
  @abstractmethod #Every concrete shape must implement area()
  def area(self): #Define the required operation
    pass #No one fits every shape
class Circle(Shape): #Circle inherits from shape
  def __init__(self,radius): #recieve the Ciricle's radius
    self.radius = radius #Store the radius
  def area(self): #circle's implementation of area()

    return 3.14*self.radius*self.radius #pi x r x r
class Square(Shape): #Square inherits from shape
  def __init__(self,side): #Recieve the squares's side length
    self.side = side #store the side length
  def area(self): #square's implementation of area()

    return self.side*self.side # side x side
circle = Circle(5) #create a circle with radius 5
square = Square(4) #create a square with side 4
print(circle.area()) #calculate and print circle's area
print(square.area())    #calculate and print square's area
```

THREAD

Thread is a path of work that a program can execute

SEQUENTIAL AND CONCURRENT

```python
import threading
import time
def download_file():
  for i in range(1, 4):
    print(f"Downloading chunk {i}...")
    time.sleep(0.5)
def play_music():
  for i in range(1, 4):
    print(f"Playing beat {i}...")
    time.sleep(0.5)
t1 = threading.Thread(target=download_file)
t2 = threading.Thread(target=play_music)
t1.start()
t2.start()
t1.join()
t2.join()
print("Both tasks are finihed!")


```

```python
import threading
import time
def numbers():
  for n in range(1, 6):
    print("Number:", n)
    time.sleep(0.3)
def letters():
  for letter in {"A", "B", "C", "D", "E"}:
    print("Letter:", letter)
    time.sleep(0.3)
t1 = threading.Thread(target=numbers)
t2 = threading.Thread(target=letters)
t1.start()
t2.start()
t1.join()
t2.join()
print("Done!")
```

Multiprocessing

```python
import multiprocessing
def print_num():
  for num in range(5):
    print(num)
def alpha():
  letters = ['a', 'b', 'c', 'd', 'e']
  for letter in letters:
    print(letter)
process1 = multiprocessing.Process(target=print_num)
process2 = multiprocessing.Process(target=alpha)
process1.start()
process1.join()
process2.start()
process2.join()



```

DAEMON THREAD

A daemon thread is a background thread that runs continuously to perform tasks while the main program is running.when main program ends,the daemon thread also ends automatically

```python
import threading
import time
def watch():
  while True:
    print(time.strftime("%H:%M:%S"))
    time.sleep(1)
t = threading.Thread(target=watch)
t.daemon = True
t.start()
#time.sleep(5)
print("Main program ended")
```
