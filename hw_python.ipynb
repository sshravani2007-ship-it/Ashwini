Create a function that calculates the simple interest of a property purchase!
Simple and compound

Simple Interest

```python
def simple_interest(p, r, t):
  return (p * r * t) / 100
  print(simple_interest(10000, 5, 2))

```

Compound Interest

```python
def compound_interest(p, r, t):
  return p * (1 + r/ 100) ** t-p
print(compound_interest(10000, 5, 2))
```

```python
def interest(p, r, t):
  simple = (p * r * t) / 100
  compound = p *(1 + r / 100)** t - p
  print("Simple Interest =", simple)
  print("Compound Interest =", compound)
interest(10000, 5, 2)

```

Outer and Inner Class

```python
class Outer:
  def display(self):
    print("This is outer class")

  class Inner:
    def show(self):
      print("This is inner class")
obj = Outer()
obj.display()
inner = Outer.Inner()
inner.show()


```

Single Inheritance

```python
class Parent:
  def show(self):
    print("This is parent class")
class Child(Parent):
  def display(self):
    print("This is child class")
obj = Child()
obj.show()
obj.display()
```

Multilevel Inheritence

```python
class Grandparent:
  def show1(self):
    print("Grandparent")
class Parent(Grandparent):
  def show2(self):
    print("Parent")
class Child(Parent,Grandparent):
  def show3(self):
    print("Child")
obj = Child()
obj.show1()
obj.show2()
obj.show3()

```

Abstraction

```python
from abc import ABC, abstractmethod
class Animal(ABC):
  @abstractmethod
  def sound(self):
    pass
class Dog(Animal):
  def sound(self):
    print("Dog barks")
d = Dog()
d.sound()

```

Polymorphism

```python
class Dog:
  def sound(self):
    print("Dog barks")
class Cat:
  def sound(self):
    print("Cat meows")
for animal in [Dog(), Cat()]:
  animal.sound()
```

Encapsulation

```python
class Student:
  def __init__(self):
    self .__marks = 90
  def get_marks(self):
    return self .__marks
s = Student()
print(s.get_marks())
```
